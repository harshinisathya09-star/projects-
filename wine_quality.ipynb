```python
import matplotlib.pyplot as plt
plt.boxplot(df["alcohol"])
plt.title("Alcohol Box Plot")
plt.show()
```

```python
Q1 = df["alcohol"].quantile(0.25)
Q3 = df["alcohol"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df = df[(df["alcohol"] >= lower) & (df["alcohol"] <= upper)]
```

```python
good_wines = df[df["quality"] >= 7]
good_wines
```

```python
df.sort_values(by="alcohol", ascending=False)
```

```python
df["quality"].hist(bins=10)
plt.title("Wine Quality Distribution")
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()
```

```python
df["quality"].value_counts().sort_index().plot(kind="bar")
plt.title("Wine Quality Count")
plt.xlabel("Quality")
plt.ylabel("Number of Wines")
plt.show()
```

```python
plt.scatter(df["alcohol"], df["quality"])
plt.title("Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.show()
```
