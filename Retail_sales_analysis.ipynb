```python
print(df["Sales"].min())
```

```python
region_data = df.groupby("Region")[["Sales","Profit"]].sum()
print(region_data)
```

```python
category_data = df.groupby("Category")[["Sales","Profit"]].sum()
print(category_data)
```

```python
df["Category"].value_counts().plot(kind="bar")
plt.title("Category Count")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
region_data["Sales"].plot(kind="line", marker="o")
plt.title("Region-wise Sales")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
df["Sales"].hist(bins=20)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Count")
plt.show()
```

```python
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```
